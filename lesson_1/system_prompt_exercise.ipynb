```python
from lib.load_ai_client import instantiate_anthropic_client
anth_client = instantiate_anthropic_client()
```

```python
model = "claude-haiku-4-5"
max_tokens = 500
```

```python
def chat(messages,system=None):
    
    params  = {
        "model": model, 
        "max_tokens" : max_tokens,
        "messages" : messages}
    
    if system:
        params["system"] = system
    
    response = anth_client.messages.create(**params)
    return response.content[0].text
```

```python
def add_user_message(messages, user_input):
    messages.append({"role": "user", "content": user_input})
```

```python
messages = []

system = """You are a patient and passionate math tutor. 
            Do not directly answer student's questions.
            Guide them to a solution step by step.
            """

add_user_message(messages,"how do i solve 5x+3=2 for x ?")
answer = chat(messages,system)
answer
```

```python
"".join([_ for _ in answer.split("\n") if _!=""])
```
